```python
import pandas as pd
import numpy as np
from statsmodels.stats.inter_rater import fleiss_kappa
from scipy.stats import norm
```

# Preparation

```python
new1 = pd.read_csv('new1.csv')
new2 = pd.read_csv('new2.csv')
new3 = pd.read_csv('new3.csv')
old1 = pd.read_csv('old1.csv')
old2 = pd.read_csv('old2.csv')
old3 = pd.read_csv('old3.csv')
```

```python
new1.drop(columns=['комментарий'], inplace=True)
new1["tag"] = new1["tag"].str.lower().str.replace("+", " ", regex=False)
new1
```

```python
new2.drop(columns=['Обоснование'], inplace=True)
new2["tag"] = new2["tag"].str.lower().str.replace("+", " ", regex=False)
new2
```

```python
new3.drop(columns=['обоснование'], inplace=True)
new3["tag"] = new3["tag"].str.lower().str.replace("+", " ", regex=False)
new3
```

```python

old1.drop(columns=['Комментарии', 'комментарии еще будут дополняться'], inplace=True)
old1['tag'] = old1['tag'].str.lower().str.replace("+", " ", regex=False)
old1
```

```python
old2.drop(columns=['Другой тег', 'Доп. тег'], inplace=True)
old2['tag'] = old2['tag'].str.lower()
old2
```

```python
old3.drop(columns=['комментарий'], inplace=True)
old3['tag'] = old3['tag'].str.lower().str.replace("+", " ", regex=False)
old3
```

# Analysis

## Old

```python
merged_old = pd.merge(old1[['quote', 'tag']], 
                      old2[['quote', 'tag']], 
                      on="quote", 
                      suffixes=("_1", "_2"))

merged_old = pd.merge(merged_old, 
                      old3[['quote', 'tag']], 
                      on="quote")

merged_old = merged_old.rename(columns={"tag": "tag_3"})
merged_old
```

```python
all_labels = sorted(set(merged_old["tag_1"]) | set(merged_old["tag_2"]) | set(merged_old["tag_3"]))

table_old = []
for _, row in merged_old.iterrows():
    counts = [0] * len(all_labels)
    counts[all_labels.index(row["tag_1"])] += 1
    counts[all_labels.index(row["tag_2"])] += 1
    counts[all_labels.index(row["tag_2"])] += 1
    table_old.append(counts)

table_old = pd.DataFrame(table_old, columns=all_labels)

kappa_old = fleiss_kappa(table_old.values)
print("Fleiss' kappa:", kappa_old)
```

## New

```python
merged_new = pd.merge(new1[['quote', 'tag']], 
                      new2[['quote', 'tag']], 
                      on="quote", 
                      suffixes=("_1", "_2"))

merged_new = pd.merge(merged_new, 
                      new3[['quote', 'tag']], 
                      on="quote")

merged_new.columns = ["quote", "tag_1", "tag_2", "tag_3"]

merged_new
```

```python
all_labels = sorted(set(merged_new["tag_1"]) | set(merged_new["tag_2"]) | set(merged_new["tag_3"]))

table_new = []
for _, row in merged_new.iterrows():
    counts = [0] * len(all_labels)
    counts[all_labels.index(row["tag_1"])] += 1
    counts[all_labels.index(row["tag_2"])] += 1
    counts[all_labels.index(row["tag_3"])] += 1
    table_new.append(counts)

table_new = pd.DataFrame(table_new, columns=all_labels)

kappa_new = fleiss_kappa(table_new.values)
print(f"Fleiss' kappa: {kappa_new:.4f}")
```

## Statistical tests

```python
def fleiss_se(table):
    N, k = table.shape
    n = table.sum(axis=1)[0]
    P_bar = table.sum(axis=0) / (N * n)
    P_e = np.sum(P_bar**2)
    kappa = fleiss_kappa(table)
    var = (2 * (N - 1) * (P_e * (1 - P_e))) / (N * n * (n - 1) * (1 - P_e)**2)
    return np.sqrt(var)


se_old = fleiss_se(table_old)
se_new = fleiss_se(table_new)

z = (kappa_new - kappa_old) / np.sqrt(se_old**2 + se_new**2)
p_value = norm.sf(abs(z))

print(f"κ old: {kappa_old:.4f}, SE old: {se_old:.4f}")
print(f"κ new: {kappa_new:.4f}, SE new: {se_new:.4f}")
print(f"z: {z:.4f}, p-value: {p_value:.4f}")
```
